```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X,Y = load_diabetes(return_X_y=True)

print(X.shape)
print(Y.shape)
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
reg = LinearRegression()
reg.fit(X_train,Y_train)

print(reg.coef_)
print(reg.intercept_)
```

```python
y_pred_lin = reg.predict(X_test)

r2Score = r2_score(Y_test,y_pred_lin)
print(r2Score)
```

```python
class customSGDRegressor:

  def __init__(self,learning_rate=0.01,epochs=100):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef_ = None
    self.intercept_ = None

  def compute_cost(self, X_train, Y_train):
    m = X_train.shape[0]
    Y_pred = self.intercept_ + np.dot(X_train, self.coef_)
    return np.sum((Y_train - Y_pred) ** 2) / (2 * m)

  def fit(self,X_train,Y_train):
    self.intercept_ = 0
    self.coef_ = np.ones(X_train.shape[1])

    m = X_train.shape[0]

    for i in range(self.epochs):
      for j in range(m):
        idx = np.random.randint(0,m)

        #update rule for intercept/bias
        y_pred = np.dot(X_train[idx],self.coef_) + self.intercept_
        intercept_der = -(Y_train[idx] - y_pred)
        self.intercept_ -= (self.learning_rate * intercept_der)

        #update rule for coefficient/weights matrix
        coef_der = -np.dot((Y_train[idx] - y_pred),X_train[idx])
        self.coef_ -= (self.learning_rate * coef_der)

      # Optionally print cost every 10% of training
      if i % max(1, self.epochs // 10) == 0:
        cost = self.compute_cost(X_train, Y_train)
        print(f"Epoch {i:4d}: Cost {cost:.4f}")

    print(self.intercept_,self.coef_)

  def predict(self,X_test):
    return np.dot(X_test,self.coef_) + self.intercept_
```

```python
gdr = customSGDRegressor(learning_rate = 0.01,epochs = 100)

gdr.fit(X_train,Y_train)
y_pred_custom = gdr.predict(X_test)
```

```python
customR2Score = r2_score(Y_test,y_pred_custom)
print(customR2Score)
```
